# Obj Oriented Proogamming
A programming paradigm that organize software around objs where each object combines data & behaviour.
OOP is a way of writing programs by representing the real world entities as Objects.

every obj has two major things
- properties - characteristics of the obj
- actions - things the obj can do

OOP is popular bcz it help developers write code that is:
- organizes
- reusable
- maintainable
- less duplicated

## OOP concepts at a glance 
### Class
    Blueprint or template for creating object
    
### Object
    an instance of a class containing real data
### Attributes
    the data or characteristics of an object
### Methods
    the actions or behaviours an obj can perform

````python
class Teacher:
    def __init__(self, name, sub):
        self.name = name       # Attribute
        self.sub = sub         # Attribute

    def teach(self):           # Method
        print(self.name, "is teaching", self.sub)


t1 = Teacher("Neha", "AI")
t2 = Teacher("Abhi", "Science")
t3 = Teacher("Drone", "Philosophy")

print(t1.name, t1.sub)

t1.teach()
t2.teach()
t3.teach()
````

# INHERITANCE
One of the most powerful OOP features is the ability of one class to reuse the properties and methods of another class.

Parent class - 
The class that provides properties and methods.

Child class - 
The class that inherits those properties and methods.

````python
# Parent class (Base class)
class Employee:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def get_details(self):
    return f"Name: {self.name}, Salary: ${self.salary}"


# Child class (Derived class) inheriting from Employee
class Manager(Employee):

  def __init__(self, name, salary, department):
    super().__init__(name, salary)
    self.department = department

 
  def get_details(self):
    return f"Name: {self.name}, Salary: ${self.salary}, Department: {self.department}"



emp = Employee("Arjunh", 50000)
print(emp.get_details())

mgr = Manager("Bramhi", 80000, "SD")
print(mgr.get_details())
````

````python
## operator overloading 
#it lets u define how built in operator will behave when used with ur custom obj 
class Employee:

  def __init__(self, salary):
    self.salary = salary

  def __add__(self, other):
    return self.salary + other.salary


emp1 = Employee(50000)
emp2 = Employee(60000)

total_salary = emp1 + emp2


print(total_salary) 
````
