## Variables & Objects
python doesnt put value inside a variable

a variable is simply a name that refers to an obj


```python
a = 25
# a --------> obj[25]
'''
25 = object
a = name/reference
a refers to the object 25
'''
```

```python
b = a
'''
now it doesnt mean that value of a copied in the b
a --------> obj[25] <-------- b both are referring to the same obj
a and b are referring the same object means the location of obj is same'''
print(a)
print(b)
print(a is b)#true
```

```python
x = 100
y = x
z = y
print(x is z)#true
```

```python
## Objects can have different types
r = 21
print(type(r))
name = "rahul"
print(type(name))
marks = [98,87,76]
print(type(marks))
```

```python
#python is dynamically typed
x = 10
print(x)
x = "hello"
print(x)
x = [1, 2, 3]
print(x)
'''The name x can refer to objects of different types at different times.
This is one reason Python is called dynamically typed.'''
```

```python
##   == vs is

a = [1, 2, 3]
b = [1, 2, 3]

#a ---> [1, 2, 3]
#b ---> [1, 2, 3]

#contents are equal.
print(a == b)#true
#but they are different objects.
print(a is b )#false


```

```python
# list makes more sense
a = [10, 20]
b = a
#a --------> list[10, 20]<-------- b
b.append(30)
print(a)
print(b)

#b -----> LIST[10,20] WHICH IS A EQUAL BUT DIFF LIST
#compare with this
b = [10, 20]
b.append(50)
print(b)
```

```python
## id() --- give obj's identity not something python guarantees to be a ram address,
#For CPython (the standard Python implementation), that identity is typically the object's memory address.
a = [1, 2]
b = a
c = 25

print(id(a))
print(id(b))#same as a because both refering to same obj 
print(id(c))#unique identity of obj 
```

id() → unique identity of the object during its lifetime

CPython → this is usually the memory address
