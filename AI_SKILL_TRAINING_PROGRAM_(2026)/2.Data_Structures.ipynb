```python
students = ["Rahul", "Neha", "Amit","Abhi", "Rahul"]
print(students)
```

```python
students.append("Priya")
students.append("mayur") # add item at the end of the list
students.insert(1, "Arjunh")
students.pop() # remove the element from end
```

```python
print(students)
```

```python
names = ["abhi", "neha", "arjunh", "tejas", "madhav"]
print(f"nums = {names}")
line = "________________________________________________________________________________________________"
print(line)

names.append('krsn')  # adds one element at the end
print(f"after append(krsn): {names}")
print(line)

names.sort()  # sorts ascending — use sort(reverse=True) for descending
print(f"after sort(): {names}")
print(line)

names.reverse()  # reverses the list in place
print(f"after reverse(): {names}")
print(len(names))
print(line)

names.insert(2, "rahul")  # inserts "rahul" at index 2
print(f"after insert(2, 'rahul'): {names}")
print(len(names))
print(line)

names.remove("madhav")  # removes the first item that equals 2 (by value, not index)
print(f"after remove(madhav): {names}")
print(line)

names.pop()  # removes the last element
print(f"after pop(): {names}")
print(line)

print(f'index of "krsn" = {names.index("krsn")}')
print(line)

print(f"count of 'rahul' = {names.count('rahul')}")
```

# SET

- Unordered collection
- Contains unique values
- Does not allow duplicates
- Does not support indexing
- Good for fast membership checking

```python
students = {"Rahul", "Neha", "Amit", "Rahul", "Abhi"}
print(students)
```

Real examples
- Mobile numbers
- PAN numbers
- Vehicle numbers
- Student roll numbers
- Unique book titles

set - unique values

```python
##Methods
students.add("Priya")
print(students)
students.update(["Amit", "Arjunh"])
print(students)
students.remove("Neha")
print(students)
students.discard("Neha")
print(students)
students.pop()
print(students)
students.clear()
print(students)
```

## SET OPERATIONS
- Union - All unique elements from both sets.(|)
- Intersection - Only common elements.(&)
- Difference - Elements in first set but not second.(-)
- Symmetric Difference - Elements present in only one set, not both.(^)

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}

print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
```

| Feature    | List         | Tuple      | Set                 |

# DICTIONARY
- Key : value collection
- Stores data as key-value pairs
- Keys are unique
- Accesses values using keys instead of indexes
- Mutable, so values can be added, updated, or deleted
- Good for storing and retrieving labelled information quickly

```python
student = {
    "roll_no": 1001,
    "name": "Rahul",
    "age": 21,
    "department": "AI"
}
print(student["name"])
```

- Problem with indexing: Numeric indexes are meaningless numbers (like 0, 1, 2) that change completely if you add or remove an item at the beginning of the collection.
  
- Meaningful labels: Descriptive words or text keys used to name data points so you can easily understand what information is being retrieved.

```python
#UPDATING
student["phone"] = "9999999999" ## adding new key to the dic
print(student["phone"])
student["phone"] = "8787878878"# updating the key 
print(student["phone"])
```

```python
## dictionary methods
print(student.keys())

print(student.values())

print(student.items())

student.get("name") #If the key exists, it returns the value. If it does not exist, it returns None
```

## NESTED DICTIONARY
Dictionary inside another dictionary.

```python
students = {
    1001: {
        "name": "Rahul",
        "age": 21,
        "dept": "AI",
        "marks" : {"phy":98,"chem":87,"math":89,"stats":98}
    
    },
    1002: {
        "name": "Neha",
        "age": 24,
        "dept": "Computer"
    },
    1003:{
        "name" : "Abhijeet",
        "age" : 21,
        "dept" : "IT"
    }
}

print(students[1002]["name"])
print(students[1003]["dept"])

##safe acces -------> .get()
print(students.get(1001,{}).get("marks",{}).get("phy"))
```
