# Day 07: File Input/Output

## Part 1: Why Files?

Anything I do on screen — code, data, tasks — lives in RAM temporarily, and I lose it the moment the program ends or the PC restarts. That's why permanent storage (files) exists — so I can access the same data again later.

**Types of files:**
1. **Text files** — `.txt`, `.docx`, `.log`, etc.
2. **Binary files** — `.mp4`, `.mov`, `.png`, `.jpeg`, etc.


## Part 2: Opening a File

I have to open a file before reading or writing to it:
```python
f = open("file_name", "mode")
```

| Mode | Meaning |
|---|---|
| `'r'` | open for reading (default) |
| `'w'` | open for writing, truncating the file first (erases existing content) |
| `'x'` | create a new file and open it for writing |
| `'a'` | open for writing, appending to the end if the file already exists |
| `'b'` | binary mode |
| `'t'` | text mode (default) |
| `'+'` | open for updating (both reading and writing) |


### Write Mode — overwrites the file

````python
f = open("Day07_example.txt", "w")
f.write("Today is Day 7 of learning Python with Apna College!")
f.close()
````

### Append Mode — adds on without erasing what's already there

````python
f = open("Day07_example.txt", "a")
f.write("\nIt feels amazing to stay consistent. ")
f.write("\nLearning every single day gives you a natural dopamine hit—and honestly, that's the best kind of high.")
f.write("\nLet's keep this momentum going!")
f.close()
````

## Part 4: Reading a File

`.read()` with no argument reads the whole file. `.read(n)` reads just the first `n` characters.

````python
f = open("Day07_example.txt", "r")
data = f.read()
print(data)
f.close()

f = open("Day07_example.txt", "r")
data = f.read(15)  # reads the first 15 characters
print(data)
f.close()
````

`.readline()` reads one line at a time. Each read call moves the file's internal position forward, so the next call picks up from where the last one left off — it doesn't start over.

````python
f = open("Day07_example.txt", "r")

data = f.read(15)   # reads the first 15 characters
print(data)

line1 = f.readline()  # continues from character 15, reads to the end of that line
print(line1)

line2 = f.readline()  # reads the next full line
print(line2)

f.close()
````

A fresh `open()` call starts reading from the beginning again — `.readline()` on a brand-new file handle just gives me the first line.

````python
f = open("Day07_example.txt", "r")
data = f.readline()
print(data)
f.close()
````

## Part 5: The `with` Statement

`with` automatically closes the file for me once the block ends ,,even if an error happens partway through. It's the safer default over calling `.close()` manually, since it's easy to forget (like I did in the cell just above).

````python
with open("Day07_example.txt", "r") as f:
    data = f.read()
    print(data)
# the file is already closed here, automatically
````

## Part 6: Deleting a File

Deleting needs the `os` (operating system) module:
```python
import os
os.remove("file_name")
```
Worth remembering: this is permanent — there's no recycle bin for it.

````python
import os

os.remove("Day07_example.txt")
````

**Q2. Replace every occurrence of a word in `practice.txt` with another word.**

````python
def change_word(old_word, new_word):
    with open("practice.txt", "r") as f:
        data = f.read()

    new_data = data.replace(old_word, new_word)

    with open("practice.txt", "w") as f:
        f.write(new_data)

    with open("practice.txt", "r") as f:
        print(f.read())

change_word("Java", "Python")
````

**Q3. Check whether the word "learning" exists in the file.**

My first attempt used `.find()` in a truthy check:
```python
if data.find(a):
    print("found")
else:
    print("not found")
```
This is broken in two ways at once: `.find()` returns an **index**, not `True`/`False`. Index `0` (found right at the very start) is falsy in Python even though it's a real match, and `-1` (not found) is truthy even though it's *not* a match. My search for `"Learning"` (capital L) genuinely wasn't in the file — `.find()` correctly returned `-1` — but since `-1` is truthy, my code printed "found" anyway. Using the `in` operator avoids the whole problem, since it gives a real `True`/`False`.

````python
def find_word(word):
    with open("practice.txt", "r") as f:
        data = f.read()
    if word in data:
        print("found")
    else:
        print("not found")

find_word("learning")
````

**Q5. Given a file of comma-separated numbers, count how many are even.**

````python
with open("practice.txt", "w") as f:
    f.write("1,2,3,4,5,6,7,8,9,13,76,45,34,23")

with open("practice.txt", "r") as f:
    data = f.read()

nums = data.split(",")
count = 0
for n in nums:
    if int(n) % 2 == 0:
        count += 1

print(count)
````

## Key Takeaways

- `"w"` truncates (erases) the file before writing; `"a"` appends without touching what's already there.
- `.read(n)` and `.readline()` both move the file's internal position forward — repeated calls continue from where the last one stopped, they don't restart.
- Prefer `with open(...) as f:` over manual `f.close()` — it closes the file automatically, even if something errors out partway through.
- `os.remove()` permanently deletes a file — no recycle bin, no undo.
- `str.find()` returns an index, not a boolean. `0` (a match at the very start) is falsy, and `-1` (no match) is truthy — checking it directly with `if` can silently do the wrong thing. Use `substring in string` for a real yes/no check.
- When searching a file line by line, double-check the numbering — it's easy to accidentally skip the first line or start counting from the wrong place.
