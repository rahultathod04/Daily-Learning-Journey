# Day 03: Lists and Tuples


## Part 1: Lists

A list is an ordered, mutable collection of items — I can access items by position and change them after creating the list.

````python
marks = [98.6, 34.4, 23.6, 89.5, 45.0]
print(marks)
print(type(marks))
print(len(marks))
print(marks[0])
print(marks[3])
````

### Lists are mutable
Unlike strings, I can change an item in a list after creating it.

````python
names = ["rahul", "arjunh", "pratik", "mahadev"]
print(names)

names[0] = "abhi"
print(names)
````

### Lists can hold mixed data types
A single list can mix strings, numbers, floats, etc.

````python
student = ["rahul", 21, 89.5, "A+"]
print(student)
print(type(student))
````

### Slicing a list

```
 rahul  arjunh  pratik  mahadev  chirayu
   0       1       2        3        4
  -5      -4      -3       -2       -1
```


````python
names = ["rahul", "arjunh", "pratik", "mahadev", "chirayu"]

print(names[1:])    # arjunh to the end
print(names[:4])    # rahul to mahadev
print(names[1:4])   # arjunh to mahadev
print(names[-4:-2])
````

### Common list methods

Went through the basics: `append()`, `sort()`, `reverse()`, `insert()`, `remove()`, `pop()`.

One thing that tripped me up: `remove(2)` removes the **value** `2`, not the item at index `2` — easy to mix up with `insert(2, ...)`, which *does* use an index.


````python
nums = [2, 2, 3, 4, 7, 1]
print(f"nums = {nums}")

nums.append(4)  # adds one element at the end
print(f"after append(4): {nums}")

nums.sort()  # sorts ascending — use sort(reverse=True) for descending
print(f"after sort(): {nums}")

nums.reverse()  # reverses the list in place
print(f"after reverse(): {nums}")
print(len(nums))

nums.insert(2, "rahul")  # inserts "rahul" at index 2
print(f"after insert(2, 'rahul'): {nums}")
print(len(nums))

nums.remove(2)  # removes the first item that equals 2 (by value, not index)
print(f"after remove(2): {nums}")

nums.pop()  # removes the last element
print(f"after pop(): {nums}")
````

## Part 2: Tuples

A tuple is like a list, but immutable — once created, I can't change its values.

````python
scores = (65, 43, 23, 34, 65, 78, 98)
print(type(scores))
print(len(scores))
print(scores[3])

# scores[2] = 55   -> TypeError: 'tuple' object does not support item assignment
````

### Single-element tuples
A single-element tuple needs a trailing comma — without it, Python just treats the parentheses as grouping, not a tuple.

````python
single = (1,)
print(type(single))

# without the comma, this is just an int in parentheses, not a tuple
single = (1)
print(type(single))
````

### Slicing a tuple
Same rules as list slicing.

````python
names = ("rahul", "arjunh", "pratik", "mahadev", "chirayu")

print(names[1:])
print(names[:4])
print(names[1:4])
print(names[-4:-2])
````

### Tuple methods
Tuples don't have as many methods as lists since they can't be changed — mainly `index()` and `count()`.

````python
names = ("rahul", "arjunh", "pratik", "mahadev", "rahul", "rahul")
print(names)

print(f'index of "pratik" = {names.index("pratik")}')
print(f"count of 'rahul' = {names.count('rahul')}")
````

## Part 3: Practice Questions

**Q1. Ask the user for their 3 favorite movies and store them in a list.**

````python

m1 = input("Enter your 1st favorite movie: ")
m2 = input("Enter your 2nd favorite movie: ")
m3 = input("Enter your 3rd favorite movie: ")

movies = [m1, m2, m3]
print(f"Your favorite movies are: {movies}")
````

A shorter way to do the same thing using `append()`:

````python
movies = []
movies.append(input("Enter your 1st favorite movie: "))
movies.append(input("Enter your 2nd favorite movie: "))
movies.append(input("Enter your 3rd favorite movie: "))
print(f"Your favorite movies are: {movies}")
````

**Q2. Check whether a list is a palindrome (reads the same forwards and backwards).**

````python
list1 = [1, 2, 3, 4, 3, 2, 1]
list2 = [3, 2, 1, 5, 6]

reversed1 = list1.copy()
reversed1.reverse()

reversed2 = list2.copy()
reversed2.reverse()

print("Palindrome" if reversed1 == list1 else "Not a palindrome")
print("Palindrome" if reversed2 == list2 else "Not a palindrome")
````

**Q3. Count the number of students who got an A grade in this tuple.**

````python
grades = ('C', 'D', 'A', 'A', 'B', 'A', 'B', 'C', 'A')
print(f"Number of students who got an A grade: {grades.count('A')}")
````

**Q4. Sort the grades in a list.**

Note to self: `sort()` sorts the list **in place** and returns `None` — so `print(grades.sort())` would just print `None`. Sort first, then print the list on its own line.

````python
grades = ['C', 'D', 'A', 'A', 'B', 'A', 'B', 'C', 'A']
grades.sort()
print(grades)
````

## Key Takeaways

- Lists are ordered and mutable — I can change, add, or remove items after creating the list.
- Tuples are ordered but immutable — once created, the values can't change.
- A single-element tuple needs a trailing comma: `(1,)`, not `(1)`.
- Methods like `.sort()`, `.reverse()`, and `.remove()` modify the list in place and return `None` — they don't return the modified list.
- `.remove(value)` removes by value, not by index — easy to confuse with `.insert(index, value)`.
