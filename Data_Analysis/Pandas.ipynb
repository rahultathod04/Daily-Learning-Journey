# Pandas
Pandas is an open-source, high-performance data manipulation and analysis library for the Python programming language.

It provides foundational data structures—most notably 
- the DataFrame (a two-dimensional, tabular data structure similar to an Excel spreadsheet or SQL table)
-  the Series (a one-dimensional labeled array)—along with powerful tools for reading, cleaning, transforming, and analyzing structured data.

Key CapabilitiesData Cleaning: 
- Handles missing data, filters rows, and renames columns easily.
- File I/O: Reads from and writes to formats like CSV, Excel, SQL databases, and JSON.
- Data Alignment: Automatically aligns data based on custom labels or indexes.
- Reshaping: Easily pivots, melts, merges, and joins different datasets.

```python
#Import pandas
import pandas as pd 
#as pandas is build over a numpy ,, it is recomended to import numpy too
import numpy as np
```

# Series
A Pandas Series is a one-dimensional, labeled data structure that stores a sequence of values along with an index (label) for each value. It can hold data of a particular dtype and provides powerful tools for accessing, filtering, and performing operations on that data.

Series can hold mixed types, but homogeneous types are generally preferable for clean and efficient data analysis.

```python
##########series from list##########
country = ['India', 'America', 'USA', 'Nepal']

pd.Series(country)
```

```python
##Int
runs = [21,23,34,45,65]
pd.Series(runs)
```

```python
## Custom Index 
marks = [67,89,90,78,99]
sub = ['mar', 'math','sci','chem', 'eng']

pd.Series(marks, index = sub)
```

```python
## Setting a name
student = pd.Series(marks, index = sub, name='Nitish')
print(student)
```

```python
##########Series from Dic##########
marks = {
    'math' : 99,
    'sci' : 89,
    'mar' : 90,
    'geo' : 87
}

Student = pd.Series(marks, name = 'rahul ke marks')
print(Student)
```

# Series Attributes

```python
#size 
print(Student.size)

#dtype
print(Student.dtype)

#name
print(Student.name)

#is_unique
print(Student.is_unique)

print(pd.Series([1,2,3,3,4,5,6]).is_unique)

#index
print(Student.index)

#value
print(Student.values)
```
