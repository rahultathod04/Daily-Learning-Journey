### Control Flow
Control flow means controlling the order in which program statements are executed.
### Control Structures
Control strucutres are programming constructs that determine the order in which program statements are executed based on conditions or repetition.

# IF STATEMENT
The if statement is a control structure that executes a block of code only when a specified condition evaluates to true.
```python
if(age>=18):
    print("eligible for voting")
```

````python
#ASSIGNMENT
marks = 32
if(marks>40):
    print("pass")
if(marks<40):
    print("fail")
````

# IF-ELSE
The if-else statement is a decision-making control structure that executes one block when the condition is true and an alternative block when the condition is false.
```python
if(condition):
    #work if condition is true
else:
    #work if condition is false
```

````python
pin = 1234

if pin == 1234:
    print("Access Granted")
else:
    print("Access Denied")
````

# NESTED IF
One decision exists inside another decision.]
```python
if condition1:
    if condition2:
        print("...")
```

````python
age= 23
salary = 60
CIBIL = 400
if(age>=21):
    if(salary>=50):
        if(CIBIL>=750):
            print("Approve loan")
````

# ELIF
elif is used when there are multiple alternative conditions.

````python
marks = 82

if marks >= 90:
    print("Distinction")
elif marks >= 75:
    print("First Class")
elif marks >= 60:
    print("Second Class")
elif marks >= 35:
    print("Pass")
else:
    print("Fail")
````

## LOOPS INTRODUCTION
A loop is a control structure that repeatedly executes a block of code according to a specified condition or for a defined number of iterations.

- Repetition

Performing the same action multiple times.

- Iteration

One complete execution of the loop body.
Loop runs 5 times - 5 iterations

## TWO MAIN TYPES OF LOOPS

- For loop

Used when the number of repetitions is known.

- While loop

Used when repetition continues until a condition becomes false.

# FOR LOOP
A for loop is a control structure used to iterate over a sequence or iterable and execute a block of code once for each element.

- iterable is
any collection or seq that can return its elements once at a time

[List,
Tuple,
String,
Set,
Dictionary,
Range]
- iterator is a 
stateful stream of data that actively fetches the next element.
```python
Students = ["Rahul", "Neha", "Abhi", "Krsn"] # iterable

for student in Students:
    print(student) ## student is iterator

````python
students = ["Rahul", "Neha", "Abhi", "Priya"]

for student in students:
    print(student)
````

````python
# string
text = "apple"
for char in text:
    print(char)
````

````python
#Tuple
marks = (87,76,98,45,43,32,48,98,50)
fail_students=0
for mark in marks:
    if(mark<50):
        fail_students+=1
print(fail_students)
#Tuple cannot be modified, but it can be iterated over.
````

````python
#Set
colors = {"Black", "Dark Green", "Red"}

for color in colors:
    print(color)

#The loop works, but: Set order is not guaranteed..therefore the output order may vary.
````

````python
#DICTIONARY
student = {
    "name": "Rahul",
    "age": 21,
    "department": "AI"
}
for key in student:
    print(f"{key} : {student[key]}")


for key in student:
    print(key)

for value in student.values():
    print(value)
    
````

# Assignment using for loop
1. Maximum / highest value
2. Minimum / lowest value
3. Length
4. Average
5. Counting positive/negative values


````python
#count =ve and +ve values
numbers = (1,3,4,5,-3,-4,-5,6)
positive = 0
negative = 0

for number in numbers:

    if number > 0:
        positive = positive + 1

    if number < 0:
        negative = negative + 1
print(positive)
print(negative)
````

````python
# max value and min
nums = [10,23,34,75,98,34]
max = nums[0]
min = nums[0]
for i in nums:
    if i>max:
        max = i
    if i<min:
        min = i
    
print(max)
print(min)
````

````python
# length
count = 0
for i in range(len(nums)):
    count +=1
print("lenght", count)
````

````python
# find avg 
total = 0
count = 0
for i in range(len(nums)):
    total = total + nums[count]
    count +=1
print("avg :", total/count)
print("Avg :", sum(nums)/len(nums))
````

## WHILE LOOP
A while loop repeatedly executes a block of code as long as its condition remains true.

````python
#imp structure 
i = 0              #Initialization
while (i<5):       #Condition
    print(i)       # Body
    i+=1           # update
````

````python
#OTP verification program
correct_otp = 234354
count = 3

while count > 0:

    otp = int(input("Enter OTP: "))

    if otp == correct_otp:
        print("verification successful")
        break #break is used when we want to immediately stop the loop.
    else:
        count -= 1

        if count > 0:
            print(f"try again (attempts remaining: {count})")
        else:
            print("too many attempts.")
            print("Account blocked for 24 hours.")
````
