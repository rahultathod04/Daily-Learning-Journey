#### Why NumPy is More Concise
- No Loops Needed: Python lists require for loops to add numbers or run math on each item. NumPy applies math to the whole array at once.
- Built-in Math Functions: NumPy has fast, built-in functions like .mean(), .sum(), and .sqrt(). Python lists require extra code or imported modules like math or statistics.
- Broadcasting: NumPy automatically expands smaller arrays or single numbers (scalars) to match larger arrays during operations.

 # Broadcasting
 The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.
The smaller array is "broadcast" across the larger array so that they have compatible shapes.

```python
#SameShape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a)
print(b)
```

```python
a+b # The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations. 
```

### Broadcasting Rules

- Rule 1 (Dimension Padding): If the two arrays differ in their total number of dimensions, the shape of the array with fewer dimensions is padded with ones on its leading (left) side.

- Rule 2 (Compatibility): Two dimensions are compatible if they are equal, or if one of them is 1. If a dimension is 1, that array behaves as if it is stretched or copied along that dimension to match the larger size

- Rule 3 (Failure Condition): If in any dimension the sizes disagree and neither is equal to 1, broadcasting fails and NumPy raises a ValueError


```python
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a)
print(b)

print(a+b)
```

```python
a = np.arange(3).reshape(3,1)
b = np.arange(3).reshape(1,3)
print(a)
print(b)
## broadcasting try to match shapes so it will convert 3 - (3,3),, but hear we need (3,4)
print(a+b)
```

```python
# NumPy only stretches dimensions that have a size of exactly 1.
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(12).reshape(4,3)
print(a+b)
```

### Sigmoid 
#The sigmoid function is a mathematical formula that takes any real number and squashes it into a value between 0 and 1

```python
## sigmoid(x) = 1/(1+e**-x)
def Sigmoid(array):
    return 1/(1 + np.exp(-array))
a = np.arange(20)

Sigmoid(a)
```

### Mean Square Error - 
a statistical metric that measures the average squared difference between a model's predicted values and the actual observed values


```python
# MSE = 1/n * (i=1 to n ∑(actual + predicted)**2)
actual = np.random.randint(1,50,25)
predicted =  np.random.randint(1,50,25)
print(actual)
print(predicted)


def MSE(actual,predicted):
    return np.mean((actual-predicted)**2)

print(MSE(actual, predicted))
```

### Binary Cross Entropy
measures the difference between true labels and predicted probabilities in a two-class classification problem

```python
y_true = np.array([1,   0,   1,   0])
y_pred = np.array([0.9, 0.1, 0.2, 0.3]) 

def BCE(true, pred):
   return - np.mean(true*np.log(pred) + (1-true)*np.log(1-pred))

print(BCE(y_true, y_pred))
```
