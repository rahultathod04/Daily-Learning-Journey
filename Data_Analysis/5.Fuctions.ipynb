## Parameter

A variable declared in the function definition that receives data.

## Argument

The actual value passed during the function call.

````python
def welcome(name): #name is parameter
    print("Welcome", name)

welcome("Abhijeet")# rahul is Argument
````

During the function call, Python creates the local parameter variable:
name = "Abhijeet" 

parameter = variable 
argument = actual Value

````python
# multiple Parameters
def employee(name, employee_id, department, salary):
    print(name)
    print(employee_id)
    print(department)
    print(salary)


employee("Narayan", 178, "AI", 80000)
````

````python
# POSITIONAL ARGUMENTS - python matched argument based on there positions
def employee(name, employee_id, department, salary):
    print("Name :", name)
    print("Employee_id:" , employee_id)
    print("department :" , department)
    print("salary:", salary )

employee("rahul", 178, "Comp", 80000)
````

````python
# KEYWORD ARGUMENTS - Python matches values using parameter names, not their positions.
employee(
    employee_id=128,
    department="AI",
    salary=102000,
     name="Abhi"
)
````

````python
#DEFAULT ARGUMENTS -used when no argument is passed
def employee(name, department="IT"):
    print(name)
    print(department)

employee("arjunh")
employee("mayur", "AI")##overwriting the default para
````

````python
## BUILT-IN FUNCTIONS
#a predefined function that is embedded directly into a programming language, application, or software tool.

#input()
name = input("enter ur name")

#print(name)
print(name)

#len()
print(len(name))

#max()
print(max(name))

#min()
print(min(name))

#sum()
numbers = [10, 20, 30]
print(sum(numbers))

#round()
# Round to 2 decimal places
print(round(3.14159, 2))  # Output: 3.14

#abs()
# Turns a negative number positive
print(abs(-21))  # Output: 21
# Positive numbers stay positive
print(abs(5.5))  # Output: 5.5
````

# RETURN VALUE
A function can return a value to the place where it was called.
```python
def add(a, b):

```

# PRINT VS RETURN
print() - Displays something on the screen.

return - Sends the value back.

````python
# fulll function flow
def add(a, b):
    result = a + b
    return result

answer = add(10, 20)

print(answer)
````
