## NumPy - Numerical Python 
It is a library designed primarily for fast neumarical computation.

At its core numpy provides a powerful data str called:

- narray - N Dimensional array

### Key Characteristics of NumPY
- The ndarray 
- Homogeneous Data
- Vectorization - eliminating the need of oxplicit loops
- Speed and Efficiency - pre-compiled c code , uses less memory than python list
- Ecosystem Foundation - it act as structural foundation for almost all major python libraries (pandas, sciPy, scikit learnn, Tensorflow)

Feature	Python Lists 📋	NumPy Arrays (ndarray) 🔢
Data Types	Heterogeneous (can mix strings, integers, and objects in one list).	Homogeneous (all elements must be the exact same data type, like all float64).
Memory Structure	Scattered (stores pointers to objects scattered across your system memory).	Contiguous (stores actual data in one single, unbroken block of memory).
Mathematical Operations	Not supported natively (e.g., list1 + list2 appends them together instead of adding the numbers).	Element-wise operations (e.g., array1 + array2 adds the corresponding numbers instantly).
Speed & Performance	Slower (requires Python to check the data type of every single item during loops).	Blazing fast (written in C; runs operations at compiled hardware speeds without loops).
Size Flexibility	Dynamic (efficiently shrinks or grows as you append or remove items).	Fixed size (changing the size requires creating a brand-new array behind the scenes).

## Array Attributes


```python
a1 = np.arange(10)
a2 = np.arange(12, dtype = float).reshape(3,4)
a3 = np.arange(8).astype(str).reshape(2, 2, 2)


print(a1)
print(a2)
print(a3)
```

```python
#ndim\ tell about the dimensions
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
#shape -tell about the rows and column
print(a1.shape) # (no. of items)
print(a2.shape)#(no. of row & columns)
print(a3.shape)#(depth, rows, column)
```

```python
#size -tell no. of items
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize -tell the size in bits
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
#dtype - tell data type
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
## Changing Data type 
a3.astype(np.int32)
```

### Stacking 

```python
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
print(a4)
print(a5)
```

```python
#Horizontal stacking
h = np.hstack((a4,a5))
print(h)
```

```python
#Vertical Stacking
v = np.vstack((a4,a5))
print(v)
```

### spliting 

```python
#Horizontal spiling
np.hsplit(h,2)
```

```python
np.vsplit(v,2)
```
