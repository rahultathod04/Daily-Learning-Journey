### Problem 1 — Will a customer cancel their subscription? (Churn)

- **Business question:** Which customers are about to churn so the retention team can reach out first?
- **Target:** `churned` → binary (1 = left, 0 = stayed) within the next 30 days
- **Features:** tenure (months), monthly usage/logins, number of support tickets, plan type, last payment status
- **Metric:** Recall / F1-score — picked this over plain accuracy because churners are usually a small % of customers, so a model that just predicts "no churn" for everyone would still look "accurate" while being useless.

### Problem 2 — What will a house sell for? (Price prediction)

- **Business question:** Estimate a fair listing price for a house.
- **Target:** `sale_price` → continuous number
- **Features:** square footage, number of bedrooms/bathrooms, location/zip code, age of house, lot size
- **Metric:** RMSE (Root Mean Squared Error) — price is a real number and I want to penalize big misses more than small ones. Might also track MAE since it's easier to explain to non-technical people ("on average we're off by $X").

### Problem 3 — Will this loan applicant default? (Credit risk)

- **Business question:** Should the bank approve this loan application?
- **Target:** `default` → binary (1 = defaulted, 0 = repaid)
- **Features:** income, existing debt, credit history length, loan amount requested, employment status
- **Metric:** Precision-Recall AUC — defaults are rare compared to successful repayments, and a false negative (approving a loan that defaults) is expensive, so recall on the default class matters a lot.

**Takeaway for myself:** every time, the metric choice depends on (a) is the target rare/imbalanced, and (b) which mistake is more costly. I was picking "accuracy" for everything before today, which apparently is a rookie mistake 😅.
