### 🔎 Findings

1. **Overtime is associated with higher attrition.** Employees working overtime left at a noticeably higher rate than those who don't — worth digging into *why* (burnout? bad scheduling?) beyond just using it as a feature.
2. **Job satisfaction has a clear negative relationship with attrition** — the lower the reported satisfaction, the higher the chance someone left. Tracks intuitively, which is reassuring.
3. **Monthly income needed cleanup** — about 7% of rows were missing this value, and 3 rows were extreme outliers likely from data entry mistakes. Both were fixed (median-fill + capping) rather than dropped, to avoid losing whole rows of otherwise-good data.
4. **One column (`exit_interview_done`) was leaking the target** and has been removed. This was the single most important catch of the week — if I'd trained a model with it in, I'd have gotten a model that looks amazing on paper and is useless in reality.

### ⚠️ Risks / things I'm not fully sure about

- This is a **synthetic dataset I generated myself**, so the "signal" I found (overtime/satisfaction → attrition) is partly because I built it in on purpose. Real data won't be this cooperative.
- Only 300 rows — small sample, so any pattern I found could easily be noise if this were real data. Would want a lot more rows before trusting these findings.
- I capped outliers instead of investigating *why* they happened. In a real job, I'd want to ask "is this a data entry error or a real senior employee's actual salary?" before deciding to cap it.
- Department is imbalanced (very few HR employees) — if department turns out to matter, the HR group's small sample size could make estimates for that group unreliable.

### 🗺️ Modeling plan for later (once I get to modeling week)

- **Target:** `attrition` (binary)
- **Candidate features:** age, monthly_income, years_at_company, job_satisfaction, distance_from_home_km, overtime, department
- **Excluded:** `employee_id` (identifier, not a feature), `exit_interview_done` (leakage)
- **Likely first model:** logistic regression as a baseline (simple, interpretable, good for a first pass on a binary target)
- **Metric:** given attrition is the minority class here, going with recall/F1 or PR-AUC over plain accuracy — same reasoning as Day 1's churn problem
- **Before modeling:** need a proper train/test split, and probably to check if `department` and `overtime` need encoding (one-hot) since they're categorical
