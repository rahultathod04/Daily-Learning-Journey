## 1. From Normal Functions to Functional Programming

A normal function is a reusable block of code.

In functional programming, the thinking is more about:

**Input -> Function/Transformation -> Output**

Example:

```text
5 -> square -> 25
```

The lecture introduces this as a different way of thinking about what transformation should be applied to data.


## 2. Functions as First-Class Objects

Python treats functions as **first-class objects**.

This means a function can:
- be assigned to a variable
- be passed as an argument to another function
- be returned from another function
- be stored in a data structure

This is the foundation for using small functions with tools such as `map`, `filter` and `sorted`.


````python
def square(x):
    return x * x

message = square

print(message(5))
print(square(5))
print(message is square)
````

### Function object vs function call

```python
square
```

refers to the function object.

```python
square(5)
```

calls the function and executes it.

This distinction is important when passing a function to another function.


````python
def square(x):
    return x * x

def apply_operation(operation, value):
    return operation(value)

print(apply_operation(square, 5))
````

## 3. Why Lambda Functions?

The lecture asks an important question:

**If normal functions already exist, why do we need lambda functions?**

A lambda is useful when:
- the operation is small
- the logic is simple
- a full named function is unnecessary
- the function is mainly being passed as a value to another function
- we want a compact expression

The important point is not that lambda is better than `def`. A normal function is clearer for complex logic.


## 4. Lambda Function

### Definition

A lambda function is a **small anonymous function** written using the `lambda` keyword, generally used for a short expression.

Basic form:

```text
lambda arguments: expression
```

The expression result is returned automatically.


````python
square = lambda x: x * x

print(square(5))
print(square(6))
````

Output:

```text
25
36
```

This gives the same kind of result as:

```python
def square(x):
    return x * x
```

but the lambda version is more compact.


````python
# Normal function
def square_normal(x):
    return x * x

# Lambda function
square_lambda = lambda x: x * x

print("Normal:", square_normal(5))
print("Lambda:", square_lambda(5))
````

## 5. Lambda Syntax Breakdown

Example:

```python
lambda x: x * x
```

Parts:

```text
lambda
↓
keyword

x
↓
input parameter

:
↓
separates parameter(s) from expression

x * x
↓
expression
```

The result of the expression is returned automatically.


## 6. Lambda vs Normal Function

| Normal Function | Lambda Function |
|---|---|
| Uses `def` | Uses `lambda` |
| Usually has a function name | Anonymous by design, but can be assigned to a variable |
| Can contain multiple statements | Contains one expression |
| Uses explicit `return` | Expression result is returned automatically |
| Better for complex logic | Better for small operations |
| Usually easier to maintain for larger logic | Useful for short temporary operations |


### Important point

Lambda is **not automatically better** than a normal function.

Use a normal `def` function when the logic becomes complex or when readability and maintenance matter more than compact syntax.


## 7. Celsius to Fahrenheit Example

The lecture uses Celsius-to-Fahrenheit as a simple mathematical transformation.

Formula:

```text
Fahrenheit = Celsius × 9/5 + 32
```

The same calculation can be written with a normal function and with lambda.


````python
# Normal function

def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

temperature = celsius_to_fahrenheit(25)

print("Fahrenheit:", temperature)
````

````python
# Lambda function

celsius_to_fahrenheit_lambda = lambda celsius: celsius * 9 / 5 + 32

temperature = celsius_to_fahrenheit_lambda(25)

print("Fahrenheit:", temperature)
````

Both versions perform the same transformation. The lecture output example gives `77.0` for 25°C.

## 8. Multiple Parameters in Lambda

Lambda is not restricted to one parameter.

### Addition

```python
add = lambda a, b: a + b
```

For:

```text
10 + 20
```

the result is:

```text
30
```


````python
add = lambda a, b: a + b

print(add(10, 20))
````

### Multiplication


````python
multiply = lambda a, b: a * b

print(multiply(5, 4))
````

### Three parameters

Lambda can also accept three parameters. The lecture uses finding the maximum among three values as an example.


````python
maximum = lambda a, b, c: max(a, b, c)

print(maximum(79, 65, 52))
````

The number of parameters is not the main limitation. The important limitation is that a lambda contains **one expression**, so it should stay suitable for concise logic.


## 9. Lambda Cannot Contain a Normal Multi-Statement Function Body

Valid:

```python
lambda x: x * 2
```

Also valid:

```python
lambda x: x + 10
```

But this normal multi-statement style is not valid as a lambda body:

```text
lambda x:
    y = x * 2
    return y
```

For complicated multi-step logic, use a normal `def` function instead.


## 10. Does Lambda Need `return`?

No.

The result of the lambda expression is returned automatically.

Example:

```python
square = lambda x: x * x
print(square(6))
```

The result is `36`.

There is no explicit `return` inside the lambda.


## 11. Internal Working of Lambda

Conceptually, when Python evaluates:

```python
square = lambda x: x * x
```

it creates a function object representing the lambda expression and assigns that function object to `square`.

When:

```python
square(5)
```

is executed:
1. `square` refers to the lambda function object.
2. `5` is passed as the argument.
3. `x` becomes `5`.
4. Python evaluates `x * x`.
5. The result `25` is returned automatically.


````python
square = lambda x: x * x

result_1 = square(5)
result_2 = square(6)

print(result_1)
print(result_2)
````

## 12. Lambda with `map()`

`map()` is useful when we want to **apply a function to every element** of a collection.

Lecture example:

```text
1 2 3 4 5
↓
square each value
↓
1 4 9 16 25
```

Conceptually:

```text
List
↓
Function applied to every element
↓
Transformed result
```


````python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * x, numbers))

print(result)
````

### Why `list()`?

In Python 3, `map()` returns a **map iterator**.

To get the transformed results as a list:

```python
list(map(...))
```

is used.


### Normal loop vs map

Normal loop:


````python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number * number)

print(squares)
````

Using `map()` + lambda:


````python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
````

The output is the same. The second version is a compact functional-programming approach.


## 13. Lambda with `filter()`

`filter()` is different from `map()`.

### `map()`
**Transforms** each element.

### `filter()`
**Selects** only the elements for which a condition is true.

Lecture example:

```text
1 2 3 4 5 6
↓
keep only even numbers
↓
2 4 6
```


````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````

Conceptual execution:

```text
1 -> False -> remove
2 -> True  -> keep
3 -> False -> remove
4 -> True  -> keep
5 -> False -> remove
6 -> True  -> keep
```

So `filter()` keeps the elements for which the lambda condition returns `True`.


## 14. `map()` vs `filter()`

| Tool | Main purpose |
|---|---|
| `map()` | Transform every element |
| `filter()` | Select elements satisfying a condition |

Easy memory:

```text
map    -> change
filter -> select
```


## 15. Lambda with `sorted()`

The lecture uses student records and sorting by marks.

Suppose each tuple contains:

```text
(student_name, marks)
```

To sort by marks, use a key function.


````python
students = [
    ("Rahul", 78),
    ("Priya", 93),
    ("Amit", 89),
    ("Neha", 99)
]

sorted_students = sorted(students, key=lambda student: student[1])

print(sorted_students)
````

Here:

```python
key=lambda student: student[1]
```

means:

**Use the second element of each tuple as the sorting key.**

For these tuples, the second element is the student's marks.


````python
students = [
    ("Rahul", 78),
    ("Priya", 93),
    ("Amit", 89),
    ("Neha", 99)
]

sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print(sorted_students)
````

`reverse=True` gives descending order.

The lecture points out that seeing:

```python
sorted(data, key=lambda ...)
```

is a very common real-world pattern. The lambda defines the property by which the items should be sorted.

## 16. Lambda with Conditional Expression

A lambda can contain a **conditional expression** when the logic is genuinely small.

General form:

```text
value_if_true if condition else value_if_false
```


````python
even_or_odd = lambda x: "Even" if x % 2 == 0 else "Odd"

print(even_or_odd(10))
print(even_or_odd(7))
````

For `10`:

```text
10 % 2 == 0
→ True
→ Even
```

For `7`:

```text
7 % 2 == 0
→ False
→ Odd
```


### Pass / Fail example

A student scoring below 40 can be considered fail; otherwise pass.


````python
result = lambda marks: "Pass" if marks >= 40 else "Fail"

print(result(75))
print(result(32))
````

## 17. Multiple Conditions in Lambda

Small multi-branch logic can also be written as a nested conditional expression.


````python
grade = lambda marks: (
    "A" if marks >= 80
    else "B" if marks >= 60
    else "C" if marks >= 40
    else "F"
)

print("75 ->", grade(75))
print("32 ->", grade(32))
print("85 ->", grade(85))
````

### Important professional rule

Just because something **can** be written as a lambda does not mean it **should** be written as a lambda.

When the logic becomes complicated, use a normal `def` function because it is easier to read, maintain and explain.


````python
def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "F"

for marks in [85, 75, 55, 32]:
    print(marks, "->", calculate_grade(marks))
````
