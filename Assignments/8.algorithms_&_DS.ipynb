````python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid
        elif target > nums[mid]:
            left = mid + 1
        else:
            right = mid - 1

    return -1

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
target = 4

print("Binary search index:", binary_search(numbers, target))
````

### Recursive Binary Search


````python
def binary_search_recursive(nums, left, right, target):
    if left > right:
        return -1

    mid = (left + right) // 2

    if nums[mid] == target:
        return mid
    elif target > nums[mid]:
        return binary_search_recursive(nums, mid + 1, right, target)
    else:
        return binary_search_recursive(nums, left, mid - 1, target)

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
target = 4

result = binary_search_recursive(numbers, 0, len(numbers) - 1, target)
print("Recursive binary search index:", result)
````

## 9. Search in BST


````python
def search_bst(root, target):
    if root is None or root.val == target:
        return root

    if target < root.val:
        return search_bst(root.left, target)

    return search_bst(root.right, target)

result = search_bst(root, 60)

if result is not None:
    print("Element found:", result.val)
else:
    print("Element not found")
````

## 10. BST Insertion


````python
class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, val):
        new_node = Node(val)

        if self.root is None:
            self.root = new_node
            return

        current = self.root

        while True:
            if val < current.val:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            else:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right

    def inorder(self, node):
        if node is None:
            return []
        return self.inorder(node.left) + [node.val] + self.inorder(node.right)

bst = BinarySearchTree()

for value in [50, 30, 70, 20, 40, 60, 80]:
    bst.insert(value)

print("BST inorder after insertion:", bst.inorder(bst.root))
````

## 12. AVL Tree

An AVL tree is a self-balancing Binary Search Tree.


````python
class AVLNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.height = 1

class AVLTree:
    def height(self, node):
        return node.height if node else 0

    def update_height(self, node):
        node.height = 1 + max(self.height(node.left), self.height(node.right))

    def balance_factor(self, node):
        return self.height(node.left) - self.height(node.right) if node else 0

    def rotate_right(self, y):
        x = y.left
        t2 = x.right

        x.right = y
        y.left = t2

        self.update_height(y)
        self.update_height(x)

        return x

    def rotate_left(self, x):
        y = x.right
        t2 = y.left

        y.left = x
        x.right = t2

        self.update_height(x)
        self.update_height(y)

        return y

    def insert(self, node, key):
        if node is None:
            return AVLNode(key)

        if key < node.key:
            node.left = self.insert(node.left, key)
        else:
            node.right = self.insert(node.right, key)

        self.update_height(node)

        balance = self.balance_factor(node)

        # Left Left
        if balance > 1 and key < node.left.key:
            return self.rotate_right(node)

        # Right Right
        if balance < -1 and key > node.right.key:
            return self.rotate_left(node)

        # Left Right
        if balance > 1 and key > node.left.key:
            node.left = self.rotate_left(node.left)
            return self.rotate_right(node)

        # Right Left
        if balance < -1 and key < node.right.key:
            node.right = self.rotate_right(node.right)
            return self.rotate_left(node)

        return node

    def inorder(self, node):
        if node is None:
            return []
        return self.inorder(node.left) + [node.key] + self.inorder(node.right)

avl = AVLTree()
avl_root = None

for value in [30, 20, 10, 25, 28, 40, 50]:
    avl_root = avl.insert(avl_root, value)

print("AVL inorder:", avl.inorder(avl_root))
print("AVL root:", avl_root.key)
print("AVL height:", avl.height(avl_root))
````

## 13. Sorting Algorithms

The sorting part of the module is about understanding different ways of arranging data and comparing their performance.

The main techniques practiced are:
- Bubble Sort

The built-in Python sorting methods are also useful, but the manual implementations are important for algorithmic understanding.


### Bubble Sort

Compare adjacent values and swap them when they are in the wrong order. After each pass, a larger element moves toward the end.

Worst-case: **O(n²)**.


````python
def bubble_sort(arr):
    arr = arr.copy()
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr

numbers = [64, 34, 25, 12, 22, 11, 90]
print("Bubble sort:", bubble_sort(numbers))
````

### Selection Sort

Find the smallest remaining element and place it at the current position.

Time complexity: **O(n²)**.


````python
def selection_sort(arr):
    arr = arr.copy()

    for i in range(len(arr)):
        min_index = i

        for j in range(i + 1, len(arr)):
            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr

print("Selection sort:", selection_sort(numbers))
````

### Insertion Sort

Treat the left part as sorted and insert each new element into its correct position.

Worst-case: **O(n²)**.


````python
def insertion_sort(arr):
    arr = arr.copy()

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr

print("Insertion sort:", insertion_sort(numbers))
````

### Merge Sort

Divide the data into smaller parts, sort the parts, and merge them.

Time complexity: **O(n log n)** in best, average and worst cases.


````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr.copy()

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

print("Merge sort:", merge_sort(numbers))
````

### Quick Sort

Choose a pivot, partition the data around the pivot, and recursively sort the two parts.

Best/average: **O(n log n)**  
Worst: **O(n²)**

The performance depends strongly on how the pivot divides the data.


````python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr.copy()

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

print("Quick sort:", quick_sort(numbers))
````

| Merge | O(n log n) | O(n log n) | O(n log n) |

## 14. Dynamic Programming

Dynamic Programming (DP) solves a problem by breaking it into smaller subproblems, storing their results and reusing them to avoid repeated computation.

Two important properties:
1. Overlapping subproblems
2. Optimal substructure

Two common styles:
- **Memoization:** top-down
- **Tabulation:** bottom-up

Sometimes the memory can be reduced further using a space-optimized version.


### Fibonacci - Recursive

The direct recursive solution repeatedly computes the same smaller values.


````python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

print("Fibonacci recursive:", fibonacci_recursive(10))
````

### Fibonacci - Memoization


````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci_memo(n):
    if n <= 1:
        return n
    return fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

print("Fibonacci memoization:", fibonacci_memo(10))
````

### Fibonacci - Tabulation



````python
def fibonacci_tabulation(n):
    if n <= 1:
        return n

    dp = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]

print("Fibonacci tabulation:", fibonacci_tabulation(10))
````

### Fibonacci - Space Optimized



````python
def fibonacci_space_optimized(n):
    if n <= 1:
        return n

    prev2, prev1 = 0, 1

    for _ in range(2, n + 1):
        current = prev1 + prev2
        prev2, prev1 = prev1, current

    return prev1

print("Fibonacci space optimized:", fibonacci_space_optimized(10))
````

## 15. 0/1 Knapsack

The 0/1 Knapsack problem asks for the maximum value that can be selected without exceeding a fixed capacity.

For each item, the basic choice is:
- take it
- do not take it

Each item can be selected at most once.


````python
def knapsack_01(weights, values, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight = weights[i - 1]
        value = values[i - 1]

        for cap in range(capacity + 1):
            if weight <= cap:
                dp[i][cap] = max(
                    dp[i - 1][cap],
                    value + dp[i - 1][cap - weight]
                )
            else:
                dp[i][cap] = dp[i - 1][cap]

    return dp[n][capacity]

weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5

print("Maximum knapsack value:", knapsack_01(weights, values, capacity))
````

## 16. Longest Common Subsequence (LCS)

LCS finds the length of the longest subsequence common to two strings while keeping the order of characters.

The task in the module is to find the **length** of the LCS.


````python
def lcs_length(a, b):
    m, n = len(a), len(b)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[m][n]

s1 = "ABCBDAB"
s2 = "BDCAB"

print("LCS length:", lcs_length(s1, s2))
````

## 17. Palindrome

A palindrome reads the same from both directions.

The lecture connects palindrome problems with the DP section and asks for a palindrome program as assignment/practical work.


````python
def is_palindrome(text):
    text = text.lower().replace(" ", "")
    return text == text[::-1]

examples = ["madam", "level", "algorithm", "Never odd or even"]

for text in examples:
    print(text, "->", is_palindrome(text))
````
