## FILE HANDLING
file handling is the process of creating, opening, reading, writing, updating & managing files so thata doata can be stored permanently & retrieve whenever require

RAM - it is temporary storage..It stores: variables, program execution data
When the program ends, that temporary data disappears.

File / Permanent Storage -
a file stores information so that it remains available even after: program closes, computer process ends, computer restarts

main ooprations
- Create
- Open
- Read
- Write
- Update
- Manage
- Close

# FILE MODES
 r - reading the existingg file 
 
w - write , overwrite existing content

a - append new data

x - creating new file exclusively

r+ - read_write

w+ - write_read + old content remove

a+ -append_read

## read mode r
reading the existingg file, if not exist FileNotFoundError

````python
file = open("students.txt", "r")
data = file.read()
print(data)
file.close()
````

````python
# write mode w
#write data into the file, but i do overwrites the existing files
file = open("students.txt", "w")
file.write("My name is Rahul Tathod\n")
file.close()
````

````python
#Append mode a
# add new data to the end the existing file
f = open("students.txt", "a")
f.write("Final year Computer Engineering student\n")
f.close()
# append never delete the older data
````

## r+, w+, a+
r+

Read + Write.

The file must already exist.

w+

Write + Read.

Existing content is erased.

a+

Append + Read.

````python
f = open("students.txt", "a+")
f.write("Interested in the Data Science Domain")
data = f.read()
print(data)
f.close()
````

## FILE POINTER
File pointer keeps track of where Python is currently reading or writing in the file.

If the first read consumes some characters, the next read starts from where the previous read stopped.


````python
# readline() reads one complete line at a time.
f = open("students.txt", "r")
line1 = f.readline()
line2 = f.readline()# perform work using te file pointer 

print(line1)# firs call
print(line2)

#readlines() reads all remaining lines and returns them as a list of strings.
lines = f.readlines()
print(lines)
file.close()
````

````python
# writelines() : Suppose you have multiple strings.,Instead of writing them individually:
students = ["Rahul\n", "Priya\n", "Amit\n"]
#writelines() does not automatically add \n.
file = open("students.txt", "w")

file.writelines(students)

file.close()
````

### close()
```python 
file.close()
```
Because closing the file:
- Ends the file connection.
- Releases the file resource.
- Helps ensure pending/buffered writes are committed.

## with
with automatically handles the closing of the file.

````python
with open("students.txt", "r") as file:
    data = file.read()
    print(data)
````
