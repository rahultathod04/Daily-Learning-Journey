## Part 1: What is Polymorphism?

Polymorphism means the same interface — an operator or method name — can behave differently depending on the object it's used with. Today I focused on one specific form of it in Python: **operator overloading**, where operators like `+` and `>` are really just shortcuts for calling a method behind the scenes, one I can define myself for my own classes.

### Implicit Operator Overloading — Python already does this for built-in types

```python
# the + operator behaves differently depending on what it's given
print(1 + 2)                  # numeric addition
print("rahul" + "tathod")     # string concatenation
print([1, 2, 3] + [4, 5, 6])  # list concatenation

# same operator, three different behaviors -- that's polymorphism in action
```

## Part 2: Operator Overloading with Dunder Methods

### Without a Dunder Method

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def show_number(self):
        print(self.real, "+", self.img, "i")

    def add(self, other):
        new_real = self.real + other.real
        new_img = self.img + other.img
        return Complex(new_real, new_img)

num1 = Complex(1, 3)
num1.show_number()

num2 = Complex(4, 5)
num2.show_number()

num3 = num1.add(num2)
num3.show_number()

# this works, but calling .add() explicitly isn't as natural as just writing num1 + num2
```

### With a Dunder Method — now `+` just works

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def show_number(self):
        print(self.real, "+", self.img, "i")

    def __add__(self, other):  # this is what + actually calls behind the scenes
        new_real = self.real + other.real
        new_img = self.img + other.img
        return Complex(new_real, new_img)

num1 = Complex(1, 3)
num1.show_number()

num2 = Complex(4, 5)
num2.show_number()

num3 = num1 + num2  # Python translates this into num1.__add__(num2) automatically
num3.show_number()
```

### Operators and Their Dunder Methods

| Expression | Meaning | Dunder method |
|---|---|---|
| `a + b` | addition | `__add__()` |
| `a - b` | subtraction | `__sub__()` |
| `a * b` | multiplication | `__mul__()` |
| `a / b` | division | `__truediv__()` |
| `a % b` | modulus | `__mod__()` |


## Part 3: Practice Questions

**Q1. Define a `Circle` class with radius `r`. Add an `area()` method and a `perimeter()` method.**

*(Small terminology catch: my first draft named the second method `Parameter` instead of `Perimeter` — an easy typo, but they're unrelated words. Also switched both method names to lowercase, matching Python's usual `snake_case` convention for methods.)*

```python
class Circle():
    def __init__(self, r):
        self.radius = r

    def area(self):
        area = (self.radius ** 2) * (22 / 7)
        print(f"Area of circle with radius {self.radius} is {area}")

    def perimeter(self):
        perimeter = self.radius * (22 / 7) * 2
        print(f"Perimeter of circle with radius {self.radius} is {perimeter}")

c1 = Circle(21)
c1.area()
c1.perimeter()

# using 22/7 as an approximation for pi here — Python's math.pi gives a more precise value if needed
```

**Q2. Define an `Employee` class with `role`, `department`, and `salary`, plus a method to show details. Then create an `Engineer` class that inherits from `Employee` and adds `name` and `age`.**

```python
class Employee():
    def __init__(self, role, dept, salary):
        self.role = role
        self.dept = dept
        self.salary = salary

    def show_details(self):
        print(f"role = {self.role}")
        print(f"department = {self.dept}")
        print(f"salary = {self.salary}")

class Engineer(Employee):
    def __init__(self, name, age):
        self.name = name
        self.age = age
        super().__init__("Engineer", "IT", "60LPA")

e1 = Employee("Accountant", "Finance", "50LPA")
e1.show_details()

eng1 = Engineer("Ayush", 23)
eng1.show_details()
```

**Q3. Define an `Order` class that supports comparing orders by price using `>`.**

*(My first version named the `__gt__` parameters `ord1, ord2` — it still worked, since Python just binds whatever the first parameter is called to the calling object, but the convention for a comparison dunder method is `(self, other)`. Kept the convention here so it reads clearly to anyone else looking at it — including future me.)*

```python
class Order():
    def __init__(self, item, price):
        self.item = item
        self.price = price

    def __gt__(self, other):
        return self.price > other.price

order1 = Order("Chips", 20)
order2 = Order("Mango", 90)
order3 = Order("Milk", 60)

print(order1 > order2)
print(order2 > order3)
print(order3 > order1)
print(order3 > order2)
print(order1 > order3)
```

## Key Takeaways

- Polymorphism means the same interface behaves differently depending on the object — operator overloading (`+`, `>`, etc. on my own classes) is one specific form of it in Python.
- Dunder methods (`__add__`, `__gt__`, etc.) are what Python's built-in operators actually call behind the scenes — defining them lets my own classes plug into normal operator syntax instead of needing a separately named method.
- Python doesn't enforce parameter names, but comparison/operator dunder methods are conventionally written as `(self, other)` — sticking to that convention keeps the code readable for anyone else (including future me).
- Method names should stay in `snake_case`, not `PascalCase` — reserve capitalization for class names.
