# Day 08: Object-Oriented Programming


## Part 1: Classes and Objects

To model real-world things in code, I use objects.

- **Class** — a blueprint or mold used to create objects.
- **Object** — a single instance of a class that holds its own data and behavior.

```python
# creating a class
class Student():
    name = "rahul"
    

# creating an object (instance)
s1 = Student()
print(s1.name)
```


### Class Attributes
An attribute defined directly inside the class body is shared by the class itself, and every object made from it can read it.

````python
class FerrariF1():
    model = "Ferrari F2004"
    engine = "Ferrari Tipo 053"

class McLarenF1():
    model = "McLaren MP4/4"
    engine = "Honda RA168E"

c1 = FerrariF1()
print(f"C1 car model is {c1.model}, engine is {c1.engine}")

c2 = McLarenF1()
print(f"C2 car model is {c2.model}, engine is {c2.engine}")
````

## Part 2: The `__init__()` Constructor

`__init__()` is a special function every class has — it runs automatically whenever a new object is created from that class. It's typically used to set up the object's initial data.

````python
class Std:
    def __init__(self):
        print(self)  # self refers to the exact object being created
        print("Added a new student")

s1 = Std()  # creating the object automatically triggers __init__
print(s1)
````

*Both `print()` calls above show the same memory address — confirming `self` really is `s1`, just referenced from inside the constructor before it's even finished setting up.*

### Instance Attributes vs. Class Attributes

Parameters passed into `__init__` and assigned with `self.x = x` become **instance attributes** — each object gets its own separate copy. That's different from a class attribute (like `model` in the F1 car example above), which is shared by every object of that class.

In the example below, `clg` is a class attribute — every `Student` object shares the same one — while `name`, `rollnum`, and `gender` are instance attributes, set individually per object through the constructor.

````python
class Student():
    clg = "NBNSTIC"  # class attribute — shared by every Student object

    # parameterized constructor with default values
    def __init__(self, name="anonymous", rollnum="anonymous", gender="anonymous"):
        self.name = name
        self.rollnum = rollnum
        self.gender = gender

s1 = Student("rahul", 21, "M")
print(f"name : {s1.name}\nrollno : {s1.rollnum}\ngender : {s1.gender}\ncollege : {s1.clg}\n")

s2 = Student("Abhijeet", 24, "M")
print(f"name : {s2.name}\nrollno : {s2.rollnum}\ngender : {s2.gender}\ncollege : {s2.clg}\n")

s3 = Student()  # no arguments -> falls back to the defaults
print(f"name : {s3.name}\nrollno : {s3.rollnum}\ngender : {s3.gender}\ncollege : {s3.clg}\n")
````

## Part 3: Methods

Methods are functions that belong to a class — together, attributes and methods make up a class:

```
class = attributes + methods
```


````python
class Person():
    def __init__(self, name):
        self.name = name

    def greet(self):  # a method — something the object can do
        print(f"Hello, {self.name}")

p1 = Person("anuradha")
p1.greet()
````

### Practice: Student Average

Create a `Student` class that takes a name and marks for 3 subjects in the constructor, then a method to print the average.

````python
class Student():
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def avg(self):
        total = 0
        n = 0
        for mark in self.marks:
            n += 1
            total += mark
        print(f"{self.name}, your average marks are {total / n}")

s1 = Student("Tonny", [90, 78, 97])
s1.avg()
````

## Part 4: Static Methods

If a method doesn't need to use `self` (it doesn't touch any instance-specific data), I can mark it `@staticmethod`. It can then be called directly on the class itself, without needing to create an object first.

````python
class Student():
    @staticmethod
    def welcome():
        print("Welcome")

Student.welcome()  # called directly on the class — no object needed
````

## Part 5: The Four Pillars of OOP

1. **Encapsulation** — bundling data and methods into one unit, and restricting direct access to hide internal state.
2. **Abstraction** — hiding complex implementation details and showing only the essential features.
3. **Inheritance** — creating new classes from existing ones to reuse code and build hierarchies.
4. **Polymorphism** — letting different objects respond to the same method call, each in their own way.


## Part 6: Practice — Bank Account

Create an `Account` class with two attributes — balance and account number — plus methods for debit, credit, and printing the balance.

````python
class Account():
    def __init__(self, balance, account_no):
        self.balance = balance
        self.account_no = account_no

    def debit(self, amount):
        self.balance -= amount
        print("Amount Debited :", amount)
        print("Total balance  :", self.get_balance())

    def credit(self, amount):
        self.balance += amount
        print("Amount Credited :", amount)
        print("Total balance   :", self.get_balance())

    def get_balance(self):
        return self.balance


a1 = Account(10000, 12345)
a1.debit(3000)
a1.credit(90000)
a1.credit(50000)
````

## Key Takeaways

- A class is a blueprint; an object is a specific instance made from it.
- `__init__()` runs automatically the moment an object is created — it's where I set up that object's starting data.
- Class attributes are shared by every object of that class; instance attributes (usually set via `self.x = x` inside `__init__`) belong to one specific object.
- Class names conventionally use PascalCase, to visually separate them from variables and functions.
- `@staticmethod` marks a method that doesn't need `self` — it can be called directly on the class, no object required.
- Avoid naming variables `sum`, `list`, `str`, etc. — they shadow Python's built-ins.
