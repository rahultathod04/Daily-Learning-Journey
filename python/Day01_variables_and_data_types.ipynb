# Day 01: Variables and Data Types

## Part 1: Variables

A variable is just a name that stores a value so I can reuse it later, instead of typing the same value everywhere in the code.

````python
name = "Rahul"
age = 21
marks = 83.5

print(name)
print(age)
print(marks)
````

### Assigning values
`=` assigns a value to a variable.
```python
x = 10
y = 20
total = x + y
```

````python
x = 10
y = 20
total = x + y
print(total)
````

### Naming rules

- Must start with a letter or an underscore (`_`)
- Can contain letters, numbers, and underscores after that
- Can't start with a number
- Can't use Python keywords (like `if`, `for`) as a variable name

```python
student_name = "Amit"
total_marks = 95
_age = 21
```

## Part 2: Data Types

| Data type | Example | Meaning |
|---|---:|---|
| `int` | `21` | Whole number |
| `float` | `83.5` | Decimal number |
| `str` | `"Rahul"` | Text |
| `bool` | `True` | True/False value |
| `NoneType` | `None` | No value stored |


````python
print(type(name))
print(type(age))
print(type(marks))
````

### Dynamic Typing

Python doesn't make me declare a type before using a variable — the same variable can hold different types at different points in the program.

````python
value = 10
print(type(value))

value = "Python"
print(type(value))
````

## Part 3: Mutable vs Immutable

**Mutable** — can be changed after creation: `list`, `set`, `dict`

**Immutable** — cannot be changed after creation: `int`, `float`, `str`, `tuple`, `bool`


## Part 4: Type Conversion

Changing one data type into another.

- **Implicit** — Python converts automatically
- **Explicit (casting)** — I convert it manually using `int()`, `float()`, `str()`, `list()`, `tuple()`, `set()`


````python
a = 2
b = 4.34
c = "2"

result = a + b
print(result)

converted = a + int(c)
print(converted)
````

Python won't add a number and a string directly — the string has to be converted first.
```python
print("5" + 2)        # error
print(int("5") + 2)   # correct
```

## Part 5: Taking Input

`input()` always returns a string, so I need to convert it if I want to do math with it.

````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print(f"Welcome {name}. Next year your age will be {age + 1}.")
````

## Part 6: Practice Programs

**Q1. Sum of two numbers**

````python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print(f"The sum of the two numbers is {num1 + num2}")
````

**Q2. Area of a square**

````python
side = int(input("Enter the side of the square: "))
print(f"Area = {side * side}")
````

**Q3. Average of two floating-point numbers**

````python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
print((a + b) / 2)
````

**Q4. Check whether `a >= b`**

````python
a = int(input("a: "))
b = int(input("b: "))
print(a >= b)
````

## Key Takeaways

- A variable is just a name pointing to a value — reassigning it changes what it points to, not the type it's "locked" into.
- `type()` tells me what kind of value a variable currently holds.
- `input()` always comes back as a string — cast it with `int()`/`float()` before doing math.
- Mutable types (`list`, `set`, `dict`) can change after creation; immutable types (`int`, `float`, `str`, `tuple`, `bool`) can't.
