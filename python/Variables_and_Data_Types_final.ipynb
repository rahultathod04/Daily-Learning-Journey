## 1. Variable

A **variable** is a name used to store a value in memory. The value can change while the program runs.

### Definition
A variable is a named storage location for data.

### Why variables matter
- They make code readable
- They help reuse values
- They allow your program to work with changing data

````python
name = "Rahul"
age = 21
marks = 83.5

print(name)
print(age)
print(marks)
````

### Variable assignment

Use `=` to assign a value to a variable.

```python
x = 10
y = 20
total = x + y
```

````python
x = 10
y = 20
total = x + y
print(total)
````

### Rules for variable names

- Must start with a letter or underscore (`_`)
- Can contain letters, numbers, and underscores
- Cannot start with a number
- Cannot use Python keywords

### Good examples
```python
student_name = "Amit"
total_marks = 95
_age = 21
```

## 2. Common Data Types

| Data type | Example | Meaning |
|---|---:|---|
| `int` | `21` | Whole number |
| `float` | `83.5` | Decimal number |
| `str` | `"Rahul"` | Text |
| `bool` | `True` | True/False value |
| `NoneType` | `None` | No value stored |

````python
print(type(name))
print(type(age))
print(type(marks))
````

### Quick definitions

- **Integer (`int`)**: a whole number without a decimal point
- **Float (`float`)**: a number with a decimal point
- **String (`str`)**: text inside quotes
- **Boolean (`bool`)**: only `True` or `False`
- **None**: represents the absence of a value

## 3. Dynamic Typing

Python is **dynamically typed**, which means a variable does not need a fixed type declaration before use.

The same variable can store different types at different times.

````python
value = 10
print(type(value))

value = "Python"
print(type(value))
````

## 4. Mutable and Immutable Types

### Mutable
A mutable object can be changed after creation.

Examples:
- `list`
- `set`
- `dict`

### Immutable
An immutable object cannot be changed after creation.

Examples:
- `int`
- `float`
- `str`
- `tuple`
- `bool`

## 5. Type Conversion

Type conversion means changing one data type into another.

### Two kinds of conversion
- **Implicit conversion**: Python changes the type automatically
- **Explicit conversion**: the programmer changes the type manually, also called **casting**

### Common conversion functions
- `int()`
- `float()`
- `str()`
- `list()`
- `tuple()`
- `set()`

````python
a = 2
b = 4.34
c = "2"

result = a + b
print(result)

converted = a + int(c)
print(converted)
````

### Important note

Python does **not** add an integer and a string directly. Convert the string first using `int()` or another suitable type conversion function.

Example:
```python
print("5" + 2)   # error
print(int("5") + 2)  # correct
```

