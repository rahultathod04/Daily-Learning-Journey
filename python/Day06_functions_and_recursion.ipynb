##  Functions?

Repeating the same logic for different inputs is a sign I should wrap it in a function instead of copy-pasting the code every time.

````python
# without a function, I'd have to repeat this block every time the numbers change
a = 5
b = 10
total = a + b
print(total)
````

## Part 1: Functions

A function is a reusable block of code that performs a specific task. I define it once, then call it with different inputs whenever I need that task done.

```
input (parameters) --> [ function ] --> output (return value)
```


### A Simple Function

````python
# named it _sum instead of sum() to avoid shadowing Python's built-in sum() function
def _sum(a, b):
    return a + b

print(_sum(4, 5))     # function call with arguments 4 and 5
print(_sum(12, 3))
print(_sum(82, 12))
````

### Functions Without a Return Value

If a function doesn't explicitly `return` something, Python automatically returns `None` — even if the function prints output along the way.

````python
def _greet():
    print("Hello, Welcome")

print(_greet())  # prints "Hello, Welcome" from inside the function, then None — because _greet() never returns anything
````

### Returning a Value

````python
def avg(a, b, c):
    result = (a + b + c) / 3
    return result

print(avg(2, 3, 4))
print(avg(78, 90, 98))
````

**Built-in vs. user-defined functions**
- Built-in: `print()`, `len()`, `type()`, `range()` — provided by Python
- User-defined: functions I write myself to perform a specific task


````python
# print() itself takes a few useful arguments
print("rahul", "tathod")        # multiple values, space-separated by default
print("rahul", end=" ")         # end="" replaces the default newline
print("tathod")
````

### Default Parameters

Assigning a default value to a parameter means it's used automatically when no argument is passed for it.

````python
def product(a=1, b=1):
    return a * b

print(product())        # no arguments -> uses both defaults
print(product(5))       # only a is given -> b still defaults to 1
print(product(5, 4))    # both given -> defaults are ignored
````

## Part 3: Practice Questions

**Q1. Print the elements of a list on a single line.**

````python
def print_list(items):
    for i in range(len(items)):
        print(items[i], end=" ")

heroes = ["hanuman", "krishna", "ram", "narasimha"]
print_list(heroes)
````

**Q2. Find the factorial of n.**

````python
def factorial_iterative(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    print(fact)

factorial_iterative(5)
````

## Part 4: Recursion

Recursion is when a function calls itself, working toward a **base case** — a condition where it stops calling itself and just returns. Without a base case, it would keep calling itself until Python hits its recursion limit and raises an error.

Printing before the recursive call — counts down as it goes deeper:

````python
def count_down(n):
    if n == 0:
        return
    print(n)
    count_down(n - 1)

count_down(5)
````

Printing *after* the recursive call — the calls all go deep first, then print on their way back out, so it counts up instead:

````python
def count_up(n):
    if n == 0:
        return
    count_up(n - 1)
    print(n)

count_up(5)
````

### Recursive Factorial

````python
def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return fact(n - 1) * n

print(fact(5))
````

### Practice Questions (Recursion)

**Q1. Recursive function to calculate the sum of the first n natural numbers.**

````python
# named it sum_to_n instead of sum() to avoid shadowing Python's built-in sum() function
def sum_to_n(n):
    if n == 1:
        return 1
    else:
        return sum_to_n(n - 1) + n

print(sum_to_n(3))
````

**Q2. Recursive function to print all elements in a list.**

My first attempt had a bug — I set `idx = len(items)` as the very first line inside the function, which immediately overwrote the `idx` value being passed in. That meant the base case (`idx == len(items)`) was always true instantly, so the function returned right away without printing anything, no matter what I passed in.

````python
def display(items, idx=0):
    if idx == len(items):
        return
    print(items[idx], end=" ")
    display(items, idx + 1)

items = [1, 2, 3, 4, 5, 6]
display(items)
````

## Key Takeaways

- Functions exist to avoid repeating the same code — write the logic once, call it with different inputs.
- A function that doesn't explicitly `return` a value gives back `None` — printing inside a function isn't the same as returning from it.
- Returning a value instead of printing inside the function makes it more reusable, since the caller decides what to do with the result.
- Default parameters (`def f(a=1, b=1):`) let me call a function with fewer arguments than it technically takes.
- Recursion needs a base case, or it never stops — it'll keep calling itself until Python raises a `RecursionError`.
- Where I print inside a recursive function matters: printing *before* the recursive call counts down as it goes in; printing *after* counts back up as it unwinds.
- Avoid naming functions or variables `sum`, `list`, `str`, etc. — they shadow Python's built-ins.
