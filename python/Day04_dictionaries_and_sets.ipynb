# Day 04: Dictionaries and Sets


## Part 1: Dictionaries

In a dictionary, data is stored as **key:value** pairs. Dictionaries are unordered, mutable, and don't allow duplicate keys — if I reuse a key, it just overwrites the old value.

```python
info = {
    "name": "rahul",
    "age": 21,
    "is_adult": True,
    "marks": 89.5,
    "subject_set": ["math", "c++", "philosophy"],
    "city_tuple": ("pune", "mumbai", "brl"),
}
# a key can be a string, number, or tuple (anything immutable) — but not a list or dict

print(info)
print(type(info))
print(len(info))
```

### Properties

- **Unordered** — no index, so I access values by key, not position
- **Mutable** — I can change values after creation
- **No duplicate keys** — assigning to an existing key overwrites it


### Accessing a Value

```python
print(info["name"])
print(info["marks"])
```

### Updating a Value / Adding a New Key

```python
# assign a different value to an existing key
info["name"] = "rahul_tathod"

# assign to a new key to add it
info["art"] = "Dance"

print(info)
```

### Empty Dictionary

```python
empty_dict = {}
print(empty_dict)

empty_dict["name"] = "Sushant"
print(empty_dict)
```

## Nested Dictionaries

A dictionary can hold other dictionaries as values — useful for representing structured data, like students grouped by branch, each with their own subject marks.

```python
college = {
    "computer": {
        1: {"name": "rahul", "sub": {"OOP": 98, "Python": 89, "Data_sci": 96}},
        2: {"name": "arjunh", "sub": {"OOP": 56, "Python": 79, "Data_sci": 86}},
        3: {"name": "Abhi", "sub": {"OOP": 100, "Python": 99, "Data_sci": 89}},
    },
    "entc": {
        1: {"name": "Ayush", "sub": {"M1": 98, "electr": 89, "toc": 96}},
        2: {"name": "Madhu", "sub": {"M1": 100, "electr": 89, "toc": 56}},
        3: {"name": "chitra", "sub": {"M1": 49, "electr": 78, "toc": 90}},
    },
}

print(college)
```

### Dictionary Methods

| Method | What it returns |
|---|---|
| `.keys()` | all keys |
| `.values()` | all values |
| `.items()` | key-value pairs, each as a tuple |
| `.get(key)` | the value for `key`, or `None` if it doesn't exist (no error) |
| `.update(other_dict)` | merges `other_dict` in, overwriting any matching keys |


```python
print(f"college keys: {college.keys()}")
print(f"college keys as a list: {list(college.keys())}")
print(f"number of branches: {len(list(college.keys()))}")
```

```python
print(f"college values: {college.values()}")
print(f"college values as a list: {list(college.values())}")
```

```python
pairs = list(college.items())  # each entry comes back as a (key, value) tuple
print(college.items())
print(pairs[0])
```

`.get()` is safer than `info["key"]` when I'm not sure the key exists — it returns `None` instead of raising a `KeyError`.

```python
print(info["name"])           # works fine, key exists
print(info.get("surname"))    # key doesn't exist -> None, no error
```

`.update()` modifies the dictionary in place and returns `None` — same pattern as `.sort()` on a list. If I print its return value directly, I just get `None`; I need to print the dictionary afterward instead.

```python
new_info = {"gender": "male"}
info.update(new_info)
print(info)

info.update({"name": "Neha"})  # overwrites the existing "name" key
print(info)
```

## Part 2: Sets

A set is an unordered collection of **unique**, **immutable** elements. `int`, `float`, `str`, and `tuple` can all go in a set — but `list` and `dict` can't, since they're mutable and can't be hashed.

```python
collection = {1, 2, 3, 3, 4, 5, "rahul", "abhi", "rahul"}  # duplicates get dropped automatically

print(collection)
print(len(collection))
print(type(collection))
```

### Empty Set vs. Empty Dict — a gotcha

`{}` always creates an empty **dictionary**, never a set. To create an empty set, I have to use `set()`.

```python
empty_dict = {}
print(type(empty_dict))   # dict

empty_set = set()
print(type(empty_set))    # set
```

### Set Methods

| Method | What it does |
|---|---|
| `.add(x)` | adds `x` (no effect if it's already there) |
| `.remove(x)` | removes `x` — raises an error if `x` isn't in the set |
| `.clear()` | empties the set |
| `.pop()` | removes and returns a random element (sets are unordered, so there's no "first" one) |
| `.union(other)` | all elements from both sets, duplicates removed |
| `.intersection(other)` | only the elements common to both sets |


```python
sample_set = set()

sample_set.add(1)
sample_set.add(2)
sample_set.add(2)       # no effect — 2 is already in the set
sample_set.add("rahul")
print(sample_set)

sample_set.remove(1)
sample_set.remove("rahul")
# sample_set.remove(5)  # would raise a KeyError — 5 was never in the set
print(sample_set)

sample_set.clear()
print(sample_set)
```

```python
cities = {"pune", "mumbai", "kolkata", "hello", "oikava"}
print(cities)

cities.pop()  # removes a random element — order isn't guaranteed
print(cities)
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5}

print(set1.union(set2))         # everything from both, no duplicates
print(set1.intersection(set2))  # only what's in both
```

**Q4. Figure out a way to store `9` and `9.0` as two separate values in a set.**

A plain set won't do this — `9 == 9.0` and they hash the same, so the set collapses them into one element. The workaround is to wrap each value in something that *isn't* equal just because the numbers match, like a `(type_label, value)` tuple.

```python
nums = {9, 9.0}
print(nums)  # collapses to one element — 9 and 9.0 are treated as equal

tagged = {("int", 9), ("float", 9.0)}
print(tagged)  # both survive — the tuples themselves aren't equal, even though 9 == 9.0
```

## Key Takeaways

- Dictionary keys must be immutable (`str`, `int`, `tuple`, etc.) — lists and dicts can't be keys.
- `{}` always makes an empty **dictionary**, even though `{1, 2, 3}` (with items inside) makes a set. Use `set()` for an empty set.
- Methods that modify in place — `.update()`, `.add()`, `.sort()` — return `None`. Print the object *after* calling them, not their return value.
- A set only stores immutable elements, and uniqueness is based on value + hash, not type — `9` and `9.0` count as the same element unless I wrap them in something (like a tuple) that makes them genuinely different.
- `.get()` is the safe way to look up a dictionary key that might not exist — it returns `None` instead of raising an error.
