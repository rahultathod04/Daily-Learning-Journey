```python
#. Search for a number x in this tuple using loop
tup = (1, 4, 9, 16, 25, 36, 49, 64,81,100)
x = int(input("Enter the searching number:"))
i=0
while i < len(tup):
    if(tup[i] == x):
        print(f"present at index: {i}")
    else: print("finding...")
    i+=1
#Loop will never get end untl ti cover all the i"s in the condition
```

```python
## Break - Used to terminate loo when demand encunter
tup = (1, 4, 9, 16, 25, 36, 49, 64,81,100)
x = int(input("Enter the searching number:"))
i=0
while i < len(tup):
    if(tup[i] == x):
        print(f"present at index: {i}")
        break
    else: print("finding..")
    i+=1
```

```python
## continue -- terminate the execution in the current iteration & continues the iteration 
tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
i = 0
while i < len(tup):
    if tup[i] % 2 != 0:
        i += 1
        continue
    print(tup[i])
    i += 1
```
