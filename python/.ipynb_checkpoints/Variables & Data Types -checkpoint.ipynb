operators
an operators is a symbol that performs a certain operations betn operands 


```python
#Arithmatics operators\
a = 4
b = 6

sum = a+b
diff = b-a
mul = a*b
div = b/a
mod = b%a

print(sum)
print(diff)
print(mul)
print(div)
print(mod)
```

```python
#Relational operator 
print(a==b)#equal?
print(a!=b)#not equal?
print(a<=b)#less that equal to b
print(a>=b)#greater that equal to b
```

```python
#Assgnment operator 
num = 10 
print(num)
num = num +10
print(num)
num +=10
print(num)
num -=10
print(num)
num *=10
print(num)
num /=10
print(num)
num %=3
print(num)
num **=2
print(num)
```

```python
#Logical operator 
print(not(1))
print(not(0))

# & is true if both valuse are true 
# OR is false if both are false 
```
